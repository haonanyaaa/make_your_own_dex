# Milestone 1

The goal of the milestone1 is to create a pool contract that can receive liquidity from users and make swaps within a price range. 

## Setting
1. There will be an ETH/USDC pool contract. ETH will be (x) reserve, and USDC will be the (y) reserve.
2. We'll set the current price to 5000 USDC per 1 ETH.
3. The range we'll provide liquidity into is 4545-5500 USDC per 1 ETH.
4. We'll buy some ETH from the pool. At this point, since we have only one price range, we want the price of the trade to stay within the price range. 

```python
import math
```

## Price Calculation

Since we want the initial exchange rate to be 5000 USDC for 1 ETH, we need three ticks.
1. the current tick repersent the current price (5000 USDC for 1 ETH)
2. the lower and upper bounds of the price range we're providing liquidity into. The lower price be $4545 and the upper price be $5500.

The formula as follow:
$$ \sqrt{P} = \sqrt{\frac{y}{x}} \\  \sqrt{P_c} = \sqrt{\frac{5000}{1}} = \sqrt{5000} \approx 70.71 \\ \sqrt{P_l} = \sqrt{\frac{4545}{1}} \approx 67.42 \\ \sqrt{P_u} = \sqrt{\frac{5500}{1}} \approx 74.16 $$

The tick index will be as follows:
$$ \sqrt{P(i)} = 1.0001^i \Rightarrow i = log_{\sqrt{1.0001}} \sqrt{P(i)} $$

```python
def price_to_tick(p):
    return math.floor(math.log(p, 1.0001))

print(f"current tick: {price_to_tick(5000)}")
print(f"Lower tick: {price_to_tick(4545)}")
print(f"Upper tick: {price_to_tick(5500)}")
```

Uniswap uses Q64.96 number to store $\sqrt{P}$. this is a fixed-point number that has 64 bits for the integer part, and 96 bits for the fractional part. 

```python
# The convertion is 2 ** 96
q96 = 2 ** 96
def price_to_sqrtp(p: float):
    return int(math.sqrt(p) * q96)

print(f"current price number in q64.96: {price_to_sqrtp(5000)}")
print(f"lower price number in q64.96: {price_to_sqrtp(4545)}")
print(f"upper price number in q64.96: {price_to_sqrtp(5500)}")
```
